```python
from litellm import completion, completion_cost

response = completion(
    model="groq/llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": "What is the capital of France?"}
    ]
)

cost = completion_cost(
    completion_response=response,
    model="groq/llama-3.3-70b-versatile"
)

print("Response:", response.choices[0].message.content)

print("Cost in USD:", cost)
print("Cost in CAD:", cost * 1.35)

print("Prompt Tokens:", response.usage.prompt_tokens)
print("Completion Tokens:", response.usage.completion_tokens)
print("Total Tokens:", response.usage.total_tokens)
```

```python
# caching -dont pay twice for the same question 
# if 100 users are asking what is the capital of france, you only pay once to get the answer, then cache it and return the cached answer to the other 99 users.

import litellm

litellm.callbacks=[]
litellm.success_callbacks=[]
litellm.failure_callbacks=[]
litellm.async_success_callbacks=[]
litellm.async_failure_callbacks=[]
litellm.cache = None
print("Cache cleared:", litellm.cache)
```

```python
import litellm
import time
from litellm import completion
from litellm.caching import Cache

# Enable local cache
litellm.cache = Cache(type="local")

prompt = "What is the capital of France?"

# First call
start = time.time()

r1 = completion(
    model="groq/llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": prompt}
    ],
    caching=True
)

t1 = time.time() - start

print(f"First call : {t1:.2f}s ---> {r1.choices[0].message.content}")

# Second call
start = time.time()

r2 = completion(
    model="groq/llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": prompt}
    ],
    caching=True
)

t2 = time.time() - start

print(f"Second call: {t2:.2f}s ---> {r2.choices[0].message.content}")

# Speedup
print(f"Speedup: {t1/t2:.2f}x")
```

```python
import os
from litellm import Router

model_list = [
    {
        "model_name": "cheap-summary",
        "litellm_params": {
            "model": "openai/gpt-4o-mini",
            "max_tokens": 1000,
            "api_key": os.getenv("OPENAI_API_KEY")
        }
    },
    {
        "model_name": "coding",
        "litellm_params": {
            "model": "groq/llama-3.3-70b-versatile",
            "max_tokens": 2000,
            "api_key": os.getenv("GROQ_API_KEY")
        }
    }
]

router = Router(model_list=model_list)

# Coding request
coding_response = router.completion(
    model="coding",
    messages=[
        {
            "role": "user",
            "content": "Write a Python binary search function"
        }
    ]
)

print(coding_response.choices[0].message.content)
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatLiteLLM(
    model="groq/llama-3.3-70b-versatile"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "{question}")
])

parser = StrOutputParser()

chain = prompt | llm | parser

response = chain.invoke({
    "question": "What is machine learning?"
})

print(response)
```

```python
# task aware router - route to different models based on the task. for example, if the task is coding, route to a model that is good at coding. if the task is summarization, route to a model that is good at summarization. logs cost and latency for each request and response.

import time 
from litellm import Router, completion_cost

def classify_task(messages):
    user_message = next((m for m in messages if m["role"] == "user"), None)
    if user_message:
        content = user_message["content"].lower()
        if "code" in content or "function" in content:
            return "coding"
        elif "summarize" in content or "summary" in content:
            return "summary"
    return "general"
def smartchat(user_query:str):
    task = classify_task([{"role": "user", "content": user_query}])
    
    if task == "coding":
        model = "groq/llama-3.3-70b-versatile"
    elif task == "summary":
        model = "openai/gpt-4o-mini"
    else:
        model = "gemini/gemini-2.0-flash"

    start_time = time.time()
    
    response = completion(
        model=model,
        messages=[
            {"role": "user", "content": user_query}
        ]
    )

    latency = time.time() - start_time
    cost = completion_cost(completion_response=response, model=model)

    print(f"Task: {task}")
    print(f"Model used: {model}")
    print(f"Response: {response.choices[0].message.content}")
    print(f"Latency: {latency:.2f}s")
    print(f"Cost: ${cost:.6f}")
queries =[
    "Write a Python function to reverse a string.",
    "Summarize the plot of the movie Inception.",
    "What is the capital of France?"
]
for q in queries:
    print("\n--- New Query ---")
    smartchat(q)
```
